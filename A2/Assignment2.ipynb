Code to generate P+x, change top_k in the variable called *prediction* to change the X value

```python
from transformers import AutoTokenizer, AutoModelForCausalLM
import torch
import torch.nn.functional as F


# predict the next token function
def predict_next_tokens(model_name, prompt, top_k=7):
    tokenizer = AutoTokenizer.from_pretrained(model_name)
    model = AutoModelForCausalLM.from_pretrained(model_name)

    input_ids = tokenizer.encode(prompt, return_tensors="pt")

    with torch.no_grad():
        outputs = model(input_ids)

    logits = outputs.logits[0, -1, :]
    probabilities = F.softmax(logits, dim=-1)

    # Make sure top_k isn't larger than the vocabulary
    top_k = min(top_k, probabilities.shape[-1])

    top_k_probs, top_k_indices = torch.topk(probabilities, top_k)

    # Get the kth-highest probability token
    idx = top_k_indices[top_k - 1]
    token = tokenizer.decode([idx.item()])

    return token

# poem to change
poem = """One must have a mind of winter
To regard the frost and the boughs
Of the pine-trees crusted with snow;
And have been cold a long time
To behold the junipers shagged with ice,
The spruces rough in the distant glitter
Of the January sun; and not to think
Of any misery in the sound of the wind,
In the sound of a few leaves,
Which is the sound of the land
Full of the same wind
That is blowing in the same bare place
For the listener, who listens in the snow,
And, nothing himself, beholds
Nothing that is not there and the nothing that is."""

lines = poem.splitlines()

# Put poem together
for i, line in enumerate(lines):
    words = line.split()

    if not words:
        continue

    # Preserve punctuation
    punctuation = ""
    last_word = words[-1]

    if last_word[-1] in ",.!?;:":
        punctuation = last_word[-1]

    # Everything before the final word
    prompt = " ".join(words[:-1])

    # change to change prediction function
    prediction = predict_next_tokens(
        "openai-community/gpt2",
        prompt,
        top_k=100000 # change this to change which nth token is the replacement
    )

    # prediction
    replacement = prediction.strip()

    words[-1] = replacement + punctuation
    lines[i] = " ".join(words)

poem = "\n".join(lines)

print(poem)
```
